```python
import os, getpass
from groq import Groq

key = os.environ.get("GROQ_API_KEY") or getpass.getpass("Dán GROQ_API_KEY (ẩn khi gõ): ").strip()
client = Groq(api_key=key)
```

## Danh sách model dùng được với key của bạn

```python
ids = sorted(m.id for m in client.models.list().data)
print(f"Có {len(ids)} model. Các model Llama:")
for i in ids:
    if "llama" in i.lower():
        print("  -", i)
print("\nToàn bộ danh sách:")
for i in ids:
    print("  ", i)
```

## Thử gọi một lần
Dashboard dùng `llama-3.3-70b-versatile`, dự phòng `llama-3.1-8b-instant` (đổi trong `src/config.py`).

```python
PREFERRED = ["openai/gpt-oss-120b", "openai/gpt-oss-20b"]
model = next((m for m in PREFERRED if m in ids), None)
if model is None:
    print("Không thấy model dự kiến, chọn model khác trong danh sách và sửa src/config.py")
else:
    r = client.chat.completions.create(
        model=model, messages=[{"role": "user", "content": "Trả lời đúng một từ: OK"}],
        max_completion_tokens=300, reasoning_effort="low", temperature=0,
    )
    print(f"Thử gọi {model}:", repr(r.choices[0].message.content))
```
